```python
import geopandas as gpd
import pandas as pd
import os

# Check 0.10 degree files
print("="*80)
print("0.10 DEGREE FILES COMPARISON")
print("="*80)

# Geofeather
feather_path = 'gridded_europe/Europe_010degree.geofeather'
if os.path.exists(feather_path):
    df_010_feather = pd.read_feather(feather_path)
    print(f"\n0.10° GEOFEATHER:")
    print(f"  Shape: {df_010_feather.shape}")
    print(f"  Columns: {df_010_feather.columns.tolist()}")
    print(f"  File size: {os.path.getsize(feather_path) / 1024 / 1024:.2f} MB")
    if 'CISI' in df_010_feather.columns:
        print(f"  CISI range: [{df_010_feather['CISI'].min():.4f}, {df_010_feather['CISI'].max():.4f}]")
        print(f"  CISI mean: {df_010_feather['CISI'].mean():.4f}")
        print(f"  Non-zero CISI: {(df_010_feather['CISI'] > 0).sum():,}")

# GPKG
gpkg_path = 'gridded_europe/Europe_010degree.gpkg'
if os.path.exists(gpkg_path):
    gdf_010_gpkg = gpd.read_file(gpkg_path)
    print(f"\n0.10° GPKG:")
    print(f"  Shape: {gdf_010_gpkg.shape}")
    print(f"  Columns: {gdf_010_gpkg.columns.tolist()}")
    print(f"  File size: {os.path.getsize(gpkg_path) / 1024 / 1024:.2f} MB")
    print(f"  CRS: {gdf_010_gpkg.crs}")
    if 'CISI' in gdf_010_gpkg.columns:
        print(f"  CISI range: [{gdf_010_gpkg['CISI'].min():.4f}, {gdf_010_gpkg['CISI'].max():.4f}]")
        print(f"  CISI mean: {gdf_010_gpkg['CISI'].mean():.4f}")

# Compare 0.25 too
print("\n" + "="*80)
print("0.25 DEGREE FILES")
print("="*80)

df_025 = pd.read_feather('gridded_europe/Europe_025degree.geofeather')
print(f"\n0.25° GEOFEATHER:")
print(f"  Shape: {df_025.shape}")
if 'CISI' in df_025.columns:
    print(f"  CISI range: [{df_025['CISI'].min():.4f}, {df_025['CISI'].max():.4f}]")

print("\n" + "="*80)
print("RECOMMENDATION:")
print("  → Use GEOFEATHER (faster, smaller)")
print("  → Both formats have same data, geofeather is more efficient")
print("="*80)
```

```python
import os
import pandas as pd

print("="*80)
print("SEARCHING FOR 0.10 DEGREE CISI FILES")
print("="*80)

# Check common locations
locations = [
    'READY_data/labels/',
    'CISI_zenodo/CISI/010_degree/',
    'gridded_europe/',
    './',
]

for location in locations:
    if os.path.exists(location):
        print(f"\n{location}:")
        for file in os.listdir(location):
            if '010' in file or '0.1' in file or '10' in file.lower():
                filepath = os.path.join(location, file)
                size = os.path.getsize(filepath) / 1024 / 1024
                print(f"  {file:<50} {size:>8.2f} MB")
                
                # Try to read if it's a feather/csv
                if file.endswith('.feather'):
                    try:
                        df = pd.read_feather(filepath)
                        print(f"    → Columns: {df.columns.tolist()}")
                        if 'CISI' in df.columns:
                            print(f"    → ✓ HAS CISI DATA!")
                    except:
                        pass
```
