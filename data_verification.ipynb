# Spatial Alignment

```python
print("\n" + "="*80)
print("SPATIAL ALIGNMENT CHECK (0.25 degree files)")
print("="*80)

files_to_check = {
    'CISI': r'READY_data\labels\2024_CISI_025deg.tif',
    'GDP': r'READY_data\inputs_normalized\2019_gdp_aligned_025_normalized.tif',
    'POP': r'READY_data\inputs_normalized\2020_pop_aligned_025_normalized.tif',
    'LC': r'READY_data\landuse_onehot_025\clipped_history_2020_025_onehot.tif',
}

metadata = {}

for name, path in files_to_check.items():
    if Path(path).exists():
        with rasterio.open(path) as src:
            metadata[name] = {
                'shape': src.shape,
                'crs': src.crs,
                'transform': src.transform,
                'bounds': src.bounds,
            }

print("\nSHAPE COMPARISON:")
for name, meta in metadata.items():
    print(f"  {name:10s}: {meta['shape']}")

print("\nCRS COMPARISON:")
for name, meta in metadata.items():
    print(f"  {name:10s}: {meta['crs']}")

print("\nBOUNDS COMPARISON:")
for name, meta in metadata.items():
    bounds = meta['bounds']
    print(f"  {name:10s}: ({bounds.left:.2f}, {bounds.bottom:.2f}, {bounds.right:.2f}, {bounds.top:.2f})")

print("\nTRANSFORM COMPARISON:")
for name, meta in metadata.items():
    t = meta['transform']
    print(f"  {name:10s}: {t}")

# Check if all aligned
shapes_match = len(set(str(m['shape']) for m in metadata.values())) == 1
crs_match = len(set(str(m['crs']) for m in metadata.values())) == 1
bounds_match = len(set(str(m['bounds']) for m in metadata.values())) == 1

print(f"\nALIGNMENT STATUS:")
print(f"  All shapes match: {shapes_match}")
print(f"  All CRS match: {crs_match}")
print(f"  All bounds match: {bounds_match}")

if shapes_match and crs_match and bounds_match:
    print("\n  STATUS: All files properly aligned")
else:
    print("\n  WARNING: Files not properly aligned!")
```

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

# ============================================================================
# LOAD TRAINING FILES ONLY
# ============================================================================

print("Loading training files...")

# CISI Labels (this defines the mask)
with rasterio.open(r'READY_data\labels\2024_CISI_025deg.tif') as src:
    cisi = src.read(1)
    profile = src.profile

# GDP Normalized
with rasterio.open(r'READY_data\inputs_normalized\2019_gdp_aligned_025_normalized.tif') as src:
    gdp = src.read(1)

# Pop Normalized  
with rasterio.open(r'READY_data\inputs_normalized\2020_pop_aligned_025_normalized.tif') as src:
    pop = src.read(1)

# LC One-Hot
with rasterio.open(r'READY_data\landuse_onehot_025\clipped_history_2020_025_onehot.tif') as src:
    lc = src.read()

print(f"CISI shape: {cisi.shape}")
print(f"GDP shape: {gdp.shape}")
print(f"Pop shape: {pop.shape}")
print(f"LC shape: {lc.shape}")

# ============================================================================
# ANALYZE OVERLAP
# ============================================================================

print("\n" + "="*80)
print("OVERLAP ANALYSIS")
print("="*80)

# Create masks
cisi_valid = ~np.isnan(cisi)
gdp_valid = ~np.isnan(gdp) & np.isfinite(gdp)
pop_valid = ~np.isnan(pop) & np.isfinite(pop)
lc_valid = ~np.isnan(lc[0]) & np.isfinite(lc[0])  # Check first band

# Combined valid mask
all_valid = cisi_valid & gdp_valid & pop_valid & lc_valid

print(f"\nValid pixels:")
print(f"  CISI:        {cisi_valid.sum():,} ({cisi_valid.sum()/cisi.size*100:.1f}%)")
print(f"  GDP:         {gdp_valid.sum():,} ({gdp_valid.sum()/gdp.size*100:.1f}%)")
print(f"  Pop:         {pop_valid.sum():,} ({pop_valid.sum()/pop.size*100:.1f}%)")
print(f"  LC:          {lc_valid.sum():,} ({lc_valid.sum()/lc[0].size*100:.1f}%)")
print(f"  ALL overlap: {all_valid.sum():,} ({all_valid.sum()/cisi.size*100:.1f}%)")

# Check where CISI exists but inputs don't
cisi_only = cisi_valid & ~all_valid
inputs_only = ~cisi_valid & all_valid

print(f"\nMismatch:")
print(f"  CISI exists but inputs missing: {cisi_only.sum():,} pixels")
print(f"  Inputs exist but CISI missing:  {inputs_only.sum():,} pixels")

# ============================================================================
# OUTLIER DETECTION
# ============================================================================

print("\n" + "="*80)
print("OUTLIER DETECTION")
print("="*80)

# CISI outliers
cisi_data = cisi[cisi_valid]
cisi_p99 = np.percentile(cisi_data, 99)
cisi_p999 = np.percentile(cisi_data, 99.9)
cisi_outliers_99 = cisi > cisi_p99
cisi_outliers_999 = cisi > cisi_p999

print(f"\nCISI:")
print(f"  Min: {cisi_data.min():.4f}")
print(f"  Max: {cisi_data.max():.4f}")
print(f"  Mean: {cisi_data.mean():.4f}")
print(f"  Median: {np.median(cisi_data):.4f}")
print(f"  99th percentile: {cisi_p99:.4f}")
print(f"  99.9th percentile: {cisi_p999:.4f}")
print(f"  Outliers (>99%): {cisi_outliers_99.sum():,} pixels ({cisi_outliers_99.sum()/cisi_valid.sum()*100:.2f}%)")
print(f"  Outliers (>99.9%): {cisi_outliers_999.sum():,} pixels ({cisi_outliers_999.sum()/cisi_valid.sum()*100:.2f}%)")

# GDP outliers
gdp_data = gdp[gdp_valid]
gdp_p01 = np.percentile(gdp_data, 1)
gdp_p99 = np.percentile(gdp_data, 99)
gdp_outliers_low = (gdp < gdp_p01) & gdp_valid
gdp_outliers_high = (gdp > gdp_p99) & gdp_valid

print(f"\nGDP:")
print(f"  Min: {gdp_data.min():.4f}")
print(f"  Max: {gdp_data.max():.4f}")
print(f"  Mean: {gdp_data.mean():.4f}")
print(f"  Median: {np.median(gdp_data):.4f}")
print(f"  1st percentile: {gdp_p01:.4f}")
print(f"  99th percentile: {gdp_p99:.4f}")
print(f"  Outliers (<1%): {gdp_outliers_low.sum():,} pixels")
print(f"  Outliers (>99%): {gdp_outliers_high.sum():,} pixels")

# Pop outliers
pop_data = pop[pop_valid]
pop_p01 = np.percentile(pop_data, 1)
pop_p99 = np.percentile(pop_data, 99)
pop_outliers_low = (pop < pop_p01) & pop_valid
pop_outliers_high = (pop > pop_p99) & pop_valid

print(f"\nPOP:")
print(f"  Min: {pop_data.min():.4f}")
print(f"  Max: {pop_data.max():.4f}")
print(f"  Mean: {pop_data.mean():.4f}")
print(f"  Median: {np.median(pop_data):.4f}")
print(f"  1st percentile: {pop_p01:.4f}")
print(f"  99th percentile: {pop_p99:.4f}")
print(f"  Outliers (<1%): {pop_outliers_low.sum():,} pixels")
print(f"  Outliers (>99%): {pop_outliers_high.sum():,} pixels")

# ============================================================================
# VISUALIZE OVERLAP AND OUTLIERS
# ============================================================================

fig, axes = plt.subplots(3, 3, figsize=(18, 16))

# Row 1: Valid masks
axes[0,0].imshow(cisi_valid, cmap='RdYlGn')
axes[0,0].set_title('CISI Valid Mask')

axes[0,1].imshow(all_valid, cmap='RdYlGn')
axes[0,1].set_title('ALL Inputs Valid (Overlap)')

axes[0,2].imshow(cisi_only, cmap='Reds')
axes[0,2].set_title(f'CISI Only\n({cisi_only.sum():,} pixels)')

# Row 2: CISI analysis
im1 = axes[1,0].imshow(cisi, cmap='RdYlGn', vmin=0, vmax=0.4)
axes[1,0].set_title('CISI Full Range')
plt.colorbar(im1, ax=axes[1,0])

axes[1,1].imshow(cisi_outliers_99, cmap='Reds')
axes[1,1].set_title(f'CISI Outliers (>99%ile={cisi_p99:.3f})\n{cisi_outliers_99.sum():,} pixels')

axes[1,2].imshow(cisi_outliers_999, cmap='Reds')
axes[1,2].set_title(f'CISI Extreme Outliers (>99.9%ile={cisi_p999:.3f})\n{cisi_outliers_999.sum():,} pixels')

# Row 3: Input outliers
outlier_map_gdp = np.zeros_like(gdp)
outlier_map_gdp[gdp_outliers_low] = -1
outlier_map_gdp[gdp_outliers_high] = 1
outlier_map_gdp[~gdp_valid] = np.nan

outlier_map_pop = np.zeros_like(pop)
outlier_map_pop[pop_outliers_low] = -1
outlier_map_pop[pop_outliers_high] = 1
outlier_map_pop[~pop_valid] = np.nan

from matplotlib.colors import ListedColormap, BoundaryNorm
cmap_outliers = ListedColormap(['blue', 'gray', 'red'])
norm_outliers = BoundaryNorm([-1.5, -0.5, 0.5, 1.5], cmap_outliers.N)

im3 = axes[2,0].imshow(outlier_map_gdp, cmap=cmap_outliers, norm=norm_outliers)
axes[2,0].set_title(f'GDP Outliers\nBlue=<1%ile, Red=>99%ile')

im4 = axes[2,1].imshow(outlier_map_pop, cmap=cmap_outliers, norm=norm_outliers)
axes[2,1].set_title(f'Pop Outliers\nBlue=<1%ile, Red=>99%ile')

# Combined outliers (any input)
any_outlier = gdp_outliers_low | gdp_outliers_high | pop_outliers_low | pop_outliers_high
axes[2,2].imshow(any_outlier, cmap='Reds')
axes[2,2].set_title(f'Any Input Outlier\n{any_outlier.sum():,} pixels')

plt.tight_layout()
plt.show()

# ============================================================================
# RECOMMENDATION: CREATE CISI-BASED MASK
# ============================================================================

print("\n" + "="*80)
print("RECOMMENDATION")
print("="*80)

print(f"\nCurrent training mask (all_valid): {all_valid.sum():,} pixels")
print(f"CISI-based mask (cisi_valid): {cisi_valid.sum():,} pixels")
print(f"Difference: {cisi_valid.sum() - all_valid.sum():,} pixels would be excluded")

if (cisi_valid.sum() - all_valid.sum()) > 0:
    print("\nSUGGESTION: Use CISI mask as primary mask")
    print("This excludes pixels where inputs exist but CISI doesn't")
    print("Makes sense since CISI is your ground truth label")
else:
    print("\nCurrent mask is already optimal (CISI defines coverage)")

# Check if outliers should be excluded
high_cisi_outliers = (cisi > 0.5).sum()
if high_cisi_outliers > 0:
    print(f"\nWARNING: {high_cisi_outliers} pixels have CISI > 0.5")
    print("Consider capping CISI at 0.5 or investigating these values")
```

# Check if right CISI file is used

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt

print("="*80)
print("CHECKING CISI VALUE RANGES")
print("="*80)

# ============================================================================
# CHECK ALL CISI FILES
# ============================================================================

cisi_files = {
    'Your Training 0.1°': r'READY_data\labels\2024_CISI_010deg.tif',
    'Your Training 0.25°': r'READY_data\labels\2024_CISI_025deg.tif',
    'Zenodo 0.1° TIF': r'CISI_zenodo\CISI\010_degree\europe.tif',
    'Zenodo 0.25° TIF': r'CISI_zenodo\CISI\025_degree\europe.tif',
}

results = {}

for name, path in cisi_files.items():
    from pathlib import Path
    if Path(path).exists():
        with rasterio.open(path) as src:
            data = src.read(1)
            
        valid_data = data[~np.isnan(data)]
        
        results[name] = {
            'min': valid_data.min(),
            'max': valid_data.max(),
            'mean': valid_data.mean(),
            'median': np.median(valid_data),
            'p95': np.percentile(valid_data, 95),
            'p99': np.percentile(valid_data, 99),
            'p999': np.percentile(valid_data, 99.9),
            'data': valid_data
        }
        
        print(f"\n{name}:")
        print(f"  Min:    {results[name]['min']:.6f}")
        print(f"  Max:    {results[name]['max']:.6f}")
        print(f"  Mean:   {results[name]['mean']:.6f}")
        print(f"  Median: {results[name]['median']:.6f}")
        print(f"  95th percentile: {results[name]['p95']:.6f}")
        print(f"  99th percentile: {results[name]['p99']:.6f}")
        print(f"  99.9th percentile: {results[name]['p999']:.6f}")
    else:
        print(f"\n{name}: NOT FOUND")

# ============================================================================
# CHECK FEATHER FILE
# ============================================================================

print("\n" + "="*80)
print("CHECKING FEATHER FILE (Original Source)")
print("="*80)

import pandas as pd

feather_path = r'CISI_zenodo\CISI\010_degree\CISI_europe.feather'

if Path(feather_path).exists():
    cisi_feather = pd.read_feather(feather_path)
    
    if 'CISI' in cisi_feather.columns:
        feather_cisi = cisi_feather['CISI'].values
        
        print(f"\nFeather file CISI column:")
        print(f"  Min:    {feather_cisi.min():.6f}")
        print(f"  Max:    {feather_cisi.max():.6f}")
        print(f"  Mean:   {feather_cisi.mean():.6f}")
        print(f"  Median: {np.median(feather_cisi):.6f}")
        print(f"  95th percentile: {np.percentile(feather_cisi, 95):.6f}")
        print(f"  99th percentile: {np.percentile(feather_cisi, 99):.6f}")
        
        results['Feather Original'] = {
            'min': feather_cisi.min(),
            'max': feather_cisi.max(),
            'data': feather_cisi
        }
    else:
        print(f"CISI column not found in feather file")
        print(f"Available columns: {cisi_feather.columns.tolist()}")
else:
    print(f"Feather file not found")

# ============================================================================
# ANALYSIS
# ============================================================================

print("\n" + "="*80)
print("ANALYSIS")
print("="*80)

if results:
    max_values = {name: data['max'] for name, data in results.items()}
    
    print(f"\nMaximum CISI values across files:")
    for name, max_val in max_values.items():
        print(f"  {name:30s}: {max_val:.6f}")
    
    print("\n" + "-"*80)
    
    # Check if values are normalized/scaled
    all_max = max(max_values.values())
    
    if all_max < 1.0:
        print(f"\nWARNING: Maximum CISI = {all_max:.6f} (less than 1.0)")
        print("\nPossible issues:")
        print("  1. Data was incorrectly normalized/scaled")
        print("  2. Data represents a subset or different metric")
        print("  3. Original CISI calculation had different range")
        
        # Check if it looks like it's scaled to [0, 0.5] or similar
        if all_max < 0.5:
            print(f"\n  → Values appear scaled to [0, {all_max:.2f}] range")
            print(f"  → Should be scaled to [0, 1.0] for proper CISI")
        else:
            print(f"\n  → Values in [0, {all_max:.2f}] range")
            print(f"  → Close to [0, 1.0] but not quite reaching 1.0")
    
    elif all_max > 1.0:
        print(f"\nWARNING: Maximum CISI = {all_max:.6f} (greater than 1.0)")
        print("\nCISI should theoretically be bounded [0, 1]")
        print("Values > 1 suggest:")
        print("  1. Data quality issues")
        print("  2. Different scaling/normalization")
        print("  3. Outliers or errors")
    
    else:
        print(f"\nGOOD: Maximum CISI = {all_max:.6f} (approximately 1.0)")
        print("Values are in expected [0, 1] range")

# ============================================================================
# VISUALIZE DISTRIBUTIONS
# ============================================================================

if results:
    print("\n" + "="*80)
    print("VISUALIZING DISTRIBUTIONS")
    print("="*80)
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    axes = axes.flatten()
    
    for idx, (name, data_dict) in enumerate(results.items()):
        if idx < 4:
            ax = axes[idx]
            data = data_dict['data']
            
            # Histogram
            ax.hist(data, bins=100, edgecolor='black', alpha=0.7)
            
            # Add vertical lines for percentiles
            ax.axvline(data_dict['mean'], color='red', linestyle='--', linewidth=2, label='Mean')
            ax.axvline(data_dict['median'], color='green', linestyle='--', linewidth=2, label='Median')
            ax.axvline(data_dict['p99'], color='orange', linestyle='--', linewidth=1, label='99th %ile')
            ax.axvline(data_dict['max'], color='purple', linestyle='-', linewidth=2, label='Max')
            
            # Add reference line at 1.0
            ax.axvline(1.0, color='black', linestyle=':', linewidth=2, label='Expected max (1.0)')
            
            ax.set_xlabel('CISI Value')
            ax.set_ylabel('Frequency')
            ax.set_title(f'{name}\nMax: {data_dict["max"]:.4f}')
            ax.legend(fontsize=8)
            ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# ============================================================================
# RECOMMENDATION
# ============================================================================

print("\n" + "="*80)
print("RECOMMENDATION")
print("="*80)

if results:
    max_val = max(data['max'] for data in results.values())
    
    if max_val < 0.8:
        print("\nYour CISI data appears INCORRECTLY SCALED.")
        print(f"Current max: {max_val:.4f}")
        print(f"Expected max: ~1.0")
        print("\nYou should:")
        print("  1. Check the original Zenodo documentation")
        print("  2. Verify if feather file has correct range")
        print("  3. Re-create TIF files from feather if needed")
        print("  4. Or rescale: CISI_rescaled = CISI / max_value")
    
    elif max_val < 0.95:
        print("\nYour CISI data is CLOSE to correct range but not quite 1.0")
        print(f"Current max: {max_val:.4f}")
        print("\nOptions:")
        print("  1. Use as-is (close enough)")
        print("  2. Rescale to [0, 1]: CISI_rescaled = CISI / max_value")
        print("  3. Check if this represents a specific subset/region")
    
    else:
        print("\nYour CISI data appears CORRECTLY SCALED.")
        print(f"Max value: {max_val:.4f} ≈ 1.0")
        print("No action needed.")

print("\n" + "="*80)
```

# Copying Zenodo files into the labels folder 

```python
import shutil
from pathlib import Path
import rasterio
import numpy as np

print("="*80)
print("REPLACING CISI LABEL FILES WITH CORRECT ZENODO FILES")
print("="*80)

# Source files (correct)
source_010 = r'CISI_zenodo\CISI\010_degree\europe.tif'
source_025 = r'CISI_zenodo\CISI\025_degree\europe.tif'

# Destination files (your training labels)
dest_010 = r'READY_data\labels\2024_CISI_010deg.tif'
dest_025 = r'READY_data\labels\2024_CISI_025deg.tif'

# Step 1: Backup old files
print("\nStep 1: Backing up old files...")

if Path(dest_010).exists():
    backup_010 = dest_010 + '.OLD_BACKUP'
    shutil.copy(dest_010, backup_010)
    print(f"  Backed up: {backup_010}")
else:
    print(f"  No existing file: {dest_010}")

if Path(dest_025).exists():
    backup_025 = dest_025 + '.OLD_BACKUP'
    shutil.copy(dest_025, backup_025)
    print(f"  Backed up: {backup_025}")
else:
    print(f"  No existing file: {dest_025}")

# Step 2: Copy Zenodo files
print("\nStep 2: Copying Zenodo files...")

shutil.copy(source_010, dest_010)
print(f"  Copied: {source_010}")
print(f"       -> {dest_010}")

shutil.copy(source_025, dest_025)
print(f"  Copied: {source_025}")
print(f"       -> {dest_025}")

# Step 3: Verify
print("\nStep 3: Verifying new files...")

with rasterio.open(dest_010) as src:
    data_010 = src.read(1)
    max_010 = np.nanmax(data_010)
    print(f"\n  0.1° file:")
    print(f"    Shape: {data_010.shape}")
    print(f"    Max: {max_010:.4f}")
    print(f"    Status: {'CORRECT' if abs(max_010 - 1.0) < 0.01 else 'WRONG'}")

with rasterio.open(dest_025) as src:
    data_025 = src.read(1)
    max_025 = np.nanmax(data_025)
    print(f"\n  0.25° file:")
    print(f"    Shape: {data_025.shape}")
    print(f"    Max: {max_025:.4f}")
    print(f"    Status: {'CORRECT' if abs(max_025 - 1.0) < 0.01 else 'WRONG'}")
```

# Inspect feather file

```python
import pandas as pd
import numpy as np

print("="*80)
print("INSPECTING FEATHER FILE")
print("="*80)

feather_path = r'CISI_zenodo\CISI\010_degree\CISI_europe.feather'

# Load feather file
df = pd.read_feather(feather_path)

print(f"\nFile: {feather_path}")
print(f"Shape: {df.shape} ({df.shape[0]:,} rows, {df.shape[1]} columns)")

# Check columns
print(f"\nColumns: {df.columns.tolist()}")

# Show first few rows
print(f"\nFirst 10 rows:")
print(df.head(10))

# Check data types
print(f"\nData types:")
print(df.dtypes)

# If CISI column exists, analyze it
if 'CISI' in df.columns:
    cisi_values = df['CISI'].values
    
    print("\n" + "="*80)
    print("CISI VALUES ANALYSIS")
    print("="*80)
    
    print(f"\nStatistics:")
    print(f"  Count: {len(cisi_values):,}")
    print(f"  Min: {cisi_values.min():.6f}")
    print(f"  Max: {cisi_values.max():.6f}")
    print(f"  Mean: {cisi_values.mean():.6f}")
    print(f"  Median: {np.median(cisi_values):.6f}")
    print(f"  Std: {cisi_values.std():.6f}")
    
    print(f"\nPercentiles:")
    print(f"  25th: {np.percentile(cisi_values, 25):.6f}")
    print(f"  50th: {np.percentile(cisi_values, 50):.6f}")
    print(f"  75th: {np.percentile(cisi_values, 75):.6f}")
    print(f"  95th: {np.percentile(cisi_values, 95):.6f}")
    print(f"  99th: {np.percentile(cisi_values, 99):.6f}")
    print(f"  99.9th: {np.percentile(cisi_values, 99.9):.6f}")
    
    # Check if values exceed 1.0
    over_one = (cisi_values > 1.0).sum()
    print(f"\nValues > 1.0: {over_one:,} ({over_one/len(cisi_values)*100:.2f}%)")
    
    if cisi_values.max() <= 1.0:
        print("\n  STATUS: Max ≤ 1.0 - CORRECT range ✓")
    else:
        print(f"\n  WARNING: Max = {cisi_values.max():.4f} > 1.0")

# Check for coordinate columns
print("\n" + "="*80)
print("GEOGRAPHIC COVERAGE")
print("="*80)

coord_cols = [col for col in df.columns if col.lower() in ['lon', 'longitude', 'x', 'lat', 'latitude', 'y']]
print(f"\nCoordinate columns found: {coord_cols}")

if len(coord_cols) >= 2:
    # Assume first is longitude, second is latitude
    lon_col = [c for c in coord_cols if 'lon' in c.lower() or 'x' in c.lower()][0]
    lat_col = [c for c in coord_cols if 'lat' in c.lower() or 'y' in c.lower()][0]
    
    print(f"\nUsing: {lon_col} (longitude), {lat_col} (latitude)")
    
    lons = df[lon_col].values
    lats = df[lat_col].values
    
    print(f"\nLongitude:")
    print(f"  Min: {lons.min():.2f}°")
    print(f"  Max: {lons.max():.2f}°")
    print(f"  Range: {lons.max() - lons.min():.2f}°")
    
    print(f"\nLatitude:")
    print(f"  Min: {lats.min():.2f}°")
    print(f"  Max: {lats.max():.2f}°")
    print(f"  Range: {lats.max() - lats.min():.2f}°")
    
    print(f"\nGeographic extent:")
    print(f"  West:  {lons.min():.2f}°")
    print(f"  East:  {lons.max():.2f}°")
    print(f"  South: {lats.min():.2f}°")
    print(f"  North: {lats.max():.2f}°")
    
    # Compare to your study area
    print("\n" + "-"*80)
    print("COMPARISON TO YOUR STUDY AREA")
    print("-"*80)
    
    import rasterio
    gdp_path = r'READY_data\inputs\2019_gdp_aligned_010.tif'
    
    with rasterio.open(gdp_path) as src:
        your_bounds = src.bounds
    
    print(f"\nYour study area bounds:")
    print(f"  West:  {your_bounds.left:.2f}°")
    print(f"  East:  {your_bounds.right:.2f}°")
    print(f"  South: {your_bounds.bottom:.2f}°")
    print(f"  North: {your_bounds.top:.2f}°")
    
    print(f"\nFeather coverage:")
    feather_covers_west = lons.min() <= your_bounds.left
    feather_covers_east = lons.max() >= your_bounds.right
    feather_covers_south = lats.min() <= your_bounds.bottom
    feather_covers_north = lats.max() >= your_bounds.top
    
    print(f"  Covers your West edge:  {feather_covers_west}")
    print(f"  Covers your East edge:  {feather_covers_east}")
    print(f"  Covers your South edge: {feather_covers_south}")
    print(f"  Covers your North edge: {feather_covers_north}")
    
    if all([feather_covers_west, feather_covers_east, feather_covers_south, feather_covers_north]):
        print("\n  STATUS: Feather covers your entire study area ✓")
    else:
        print("\n  WARNING: Feather does NOT fully cover your study area")

print("\n" + "="*80)
print("CONCLUSION")
print("="*80)

print("\nThe feather file contains:")
print(f"  - {df.shape[0]:,} data points")
print(f"  - CISI values from {cisi_values.min():.4f} to {cisi_values.max():.4f}")
print(f"  - Geographic coordinates")

print("\nThis is the SOURCE data that was used to create the TIF files.")
print("The TIF files are already:")
print("  1. Rasterized (converted from points to grid)")
print("  2. Properly georeferenced")
print("  3. Ready to use")

print("\nRECOMMENDATION: Use the TIF file (already processed)")
print("  - Faster")
print("  - Less error-prone")
print("  - Same data, just in usable format")

print("\n" + "="*80)
```
